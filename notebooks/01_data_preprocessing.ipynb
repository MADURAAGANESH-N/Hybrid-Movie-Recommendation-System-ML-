```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

```python
movies = pd.read_csv("../data/raw/movies.csv")
ratings = pd.read_csv("../data/raw/ratings.csv")

print("Movies shape:", movies.shape)
print("Ratings shape:", ratings.shape)
```

```python
print("Movies missing values:")
print(movies.isnull().sum())

print("\nRatings missing values:")
print(ratings.isnull().sum())
```

```python
print("Duplicate movies:", movies.duplicated().sum())
print("Duplicate ratings:", ratings.duplicated().sum())
```

```python
movies = movies.drop_duplicates()
ratings = ratings.drop_duplicates()
```

```python
movies.to_csv(
    "../data/processed/movies_clean.csv",
    index=False
)

ratings.to_csv(
    "../data/processed/ratings_clean.csv",
    index=False
)

print("Cleaned datasets saved successfully!")
```
